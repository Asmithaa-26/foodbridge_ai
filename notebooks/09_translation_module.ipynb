```python
from transformers import MBartForConditionalGeneration, MBart50TokenizerFast
import sacrebleu
```

```python
model_name = "facebook/mbart-large-50-many-to-many-mmt"

tokenizer = MBart50TokenizerFast.from_pretrained(model_name)
model = MBartForConditionalGeneration.from_pretrained(model_name)
```

```python
tokenizer.src_lang = "en_XX"
tokenizer.tgt_lang = "ta_IN"
```

```python
def translate_text(text):

    encoded = tokenizer(text, return_tensors="pt")

    generated_tokens = model.generate(
        **encoded,
        forced_bos_token_id=tokenizer.lang_code_to_id["ta_IN"]
    )

    translation = tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)

    return translation
```

```python
input_text = ["Food is ready for distribution"]

translated = translate_text(input_text)

print("Original:", input_text)
print("Tamil Translation:", translated)
```

```python
reference = ["உணவு விநியோகத்திற்கு தயாராக உள்ளது"]

prediction = translated

bleu = sacrebleu.corpus_bleu(prediction, [reference])

print("BLEU Score:", bleu.score)
```
