```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv(r"C:\Users\asmis\OneDrive\Desktop\foodbridge\data\food_master_dataset.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["surplus_kg"], bins=40, kde=True)

plt.title("Distribution of Food Surplus (kg)")
plt.xlabel("Surplus (kg)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df["total_prepared_kg"],
    y=df["surplus_kg"]
)

plt.title("Prepared Food vs Surplus")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df["city"],
    y=df["surplus_kg"]
)

plt.title("City-wise Surplus Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df["temperature_c"],
    y=df["surplus_kg"]
)

plt.title("Temperature vs Surplus")

plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=False
)

plt.title("Feature Correlation Matrix")

plt.show()
```
